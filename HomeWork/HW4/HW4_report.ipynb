# Exercise 4L Convolutional Neural Networks Report

**Dataset:** Fashion-MNIST

## Summary
In this exercise, I designed and trained a Convolutional Neural Network to classify images from the Fashion-MNIST dataset. The goal was to accurately categorize grayscale images into one of 10 fashion classes (e.g., T-shirts, Trousers, Sneakers).

The final model achieved a test accuracy of around 93%, demonstrating that the architecture successfully learned hierarchical features from the input images. The model was trained using PyTorch with data augmentation and learning rate scheduling to ensure robust performance.

## Methodology

### Data Preparation
* **Dataset:** Fashion-MNIST, consisting of 70,000 grayscale images in 10 categories, where each image is $28\times28$ pixels.
* **Preprocessing:**
    * **Normalization:** Images were normalized with mean=(0.5,) and std=(0.5,) to scale pixel values to the range [-1, 1].
    * **Augmentation (Training only):** To prevent overfitting and increase dataset variability, I applied random horizontal flips and random rotations of 10 degrees.
    * **Batch Size:** 64.

### Model Architecture
I constructed a CNN incorporating the key components required:
1.  **Conv Block 1:** 32 filters (3x3), ReLU activation, Max Pooling (2x2).
2.  **Conv Block 2:** 64 filters (3x3), ReLU activation, Max Pooling (2x2).
3.  **Regularization:** A Dropout layer at 25% was added after flattening to randomly zero out neurons during training, reducing overfitting.
4.  **Fully Connected Layers:**
    * Dense Layer: 512 neurons (ReLU).
    * Output Layer: 10 neurons (corresponding to the 10 classes).

### Training Configuration
* **Optimizer:** Adam was chosen for its efficient convergence.
* **Loss Function:** CrossEntropyLoss was used as the standard loss function for multi-class classification.
* **Scheduler:** StepLR was implemented to reduce the learning rate by a factor of 0.1 every 5 epochs.
* **Epochs:** 10.

## Experiments & Results

### Training Performance
The training process included forward pass, loss computation, backward pass, and weight updates, with validation at the end of each epoch.

The training loss decreased steadily. The validation loss tracked closely with the training loss, stabilizing after Epoch 8 when the learning rate decayed. This close tracking indicates that the dropout and augmentation successfully prevented significant overfitting.

### Evaluation Metrics
The model was evaluated on the test set to calculate accuracy and other relevant metrics.

* **Final Test Accuracy:** **93%**
* **Precision & Recall:**
* High performance (>95%) was observed for distinct classes like **Trouser**, **Bag**, and **Ankle Boot**.
* Lower performance (~80-87%) was seen for **Shirt**, **T-shirt/top**, and **Pullover**.

### Confusion Matrix
* **Analysis:** The confusion matrix visualizes model performance across classes. It reveals that the model's primary errors are "semantic" confusions. For example:
* **Shirts** are frequently misclassified as **T-shirts** or **Coats**.
* **Sneakers** are occasionally confused with **Sandals** or **Boots**.
* Classes like **Bags** and **Trousers** have very few false positives.

## Interpretations & Visualization

### Visualizing Filters (Layer 1)
The filters in the first convolutional layer detect simple low-level features, such as diagonal edges, vertical lines, and contrast blobs. These serve as the building blocks for the network.

### Feature Maps
Feature maps were displayed after convolutional layers for sample images.

**Layer 1 Maps:** Clearly preserve the shape of the object (e.g., the outline of a T-shirt), highlighting edges.

**Layer 2 Maps:** Appear more abstract and pixelated. These maps activate on specific parts of the object (e.g., only the sleeves or the collar), proving the network is combining simple edges into complex shapes.

## Hyperparameter Tuning
I experimented with different hyperparameters to improve model performance:
1.  **Learning Rate Scheduler:** Initially, without a scheduler, the loss fluctuated in later epochs. Adding decay smoothed out the convergence.
2.  **Dropout:** Adding dropout of 25% reduced the gap between training and validation accuracy.
3.  **Data Augmentation:** Adding random rotation helped the model generalize better to slightly tilted images in the test set.

## 6. Conclusion
The CNN successfully solved the Fashion-MNIST classification task with high accuracy. The combination of convolutional layers for feature extraction and dropout for regularization resulted in a robust model. The remaining errors are largely due to the visual similarity between upper-body garments (shirts/coats/pullovers), which might be resolved by increasing the model depth or image resolution.
